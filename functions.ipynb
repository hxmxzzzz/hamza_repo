```python
## 4.ARBITRARY ARGUMENT ----> if u need to take multiple number use this argument

# 1.arbitrary positional argument:
 # function that accepts any number of positional argument without defining them in advance.these arguments are collected in tuple
```

```python
def flexible(*number): # (1,2,3)
    product=1
    print(type(number)) # tuple
    for i in number:
        product=product*i
    print(product)
    
```

```python
flexible(1,2,3)
```

```python
# 2. arbitrary keyword argument:
   # function that is accept in any type of keyword argument without defining them in advance. these argument are collected in form of dictionary
```

```python
def student(**info):
    print(info)
    print(type(info))
    for a,b in info.items():
        print(f"{a} : {b}")
```

```python
print(student(name='hamza',roll_no= 1234))  #(key,value)
```

```python
# mean function: ---> sum of all data function


def mean(*num): # (by adding * u dont need to add data type)
    total=0
    for var in num:
        total=total+var
    return total/len(num)
    
```

```python
mean(1,2,3,4,5,6)
```

```python
def min_max(*arr):
    minimum = arr[0]
    maximum = 0

    for i in arr:
        if i < minimum:
            minimum = i
        elif i > maximum:
            maximum = i

    print(minimum)
    print(maximum)
```

```python
min_max(23,24,45,93,1032)
```

```python
def build_person(first_name,last_name,age=0):
    person={"First name":first_name,"Last name":last_name,}
    if age:
        person["age"]=age
        return person
```
